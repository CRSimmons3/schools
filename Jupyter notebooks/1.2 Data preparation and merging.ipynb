### 2 Prepare population dataset

```python
df2.columns
```

Rename "All ages" column as "Total Population of LA"  
Change type of population columns to integer  
Create columns:
- "Pre-school age" counting 0-3 year olds
- "Primary age" counting 4-10 year olds
- "Secondary age" counting 11-17 year olds
- "Total under 18s" counting 0-17 year olds

```python
# Change column name
df2.rename(columns = {"All ages" : "Total Population of LA"}, inplace = True)
```

```python
# Change column types to integer
df2['0'] = df2['0'].str.replace(',', '').astype(int)
df2['1'] = df2['1'].str.replace(',', '').astype(int)
df2['2'] = df2['2'].str.replace(',', '').astype(int)
df2['3'] = df2['3'].str.replace(',', '').astype(int)
df2['4'] = df2['4'].str.replace(',', '').astype(int)
df2['5'] = df2['5'].str.replace(',', '').astype(int)
df2['6'] = df2['6'].str.replace(',', '').astype(int)
df2['7'] = df2['7'].str.replace(',', '').astype(int)
df2['8'] = df2['8'].str.replace(',', '').astype(int)
df2['9'] = df2['9'].str.replace(',', '').astype(int)
df2['10'] = df2['10'].str.replace(',', '').astype(int)
df2['11'] = df2['11'].str.replace(',', '').astype(int)
df2['12'] = df2['12'].str.replace(',', '').astype(int)
df2['13'] = df2['13'].str.replace(',', '').astype(int)
df2['14'] = df2['14'].str.replace(',', '').astype(int)
df2['15'] = df2['15'].str.replace(',', '').astype(int)
df2['16'] = df2['16'].str.replace(',', '').astype(int)
df2['17'] = df2['17'].str.replace(',', '').astype(int)
df2['Total Population of LA'] = df2['Total Population of LA'].str.replace(',', '').astype(int)
```

```python
# Create "Pre-school age" column
df2["Pre-school age"] = df2["0"] + df2["1"] + df2["2"] + df2["3"]
```

```python
# Create "Primary age" column
df2["Primary age"] = df2["4"] + df2["5"] + df2["6"] + df2["7"] + df2["8"] + df2["9"] + df2["10"]
```

```python
# Create "Secondary age" column
df2["Secondary age"] = df2["11"] + df2["12"] + df2["13"] + df2["14"] + df2["15"] + df2["16"] + df2["17"]
```

```python
# Create "Total under 18s" column
df2["Total under 18s"] = df2["Pre-school age"] + df2["Primary age"] + df2["Secondary age"]
```

```python
# List of columns to keep
columns_to_keep = ["Local authority", "Total Population of LA", "Pre-school age", "Primary age", "Secondary age", "Total under 18s"]
```

```python
df2_wrangled = df2[columns_to_keep]
```

```python
# Change name of column Name to match df1
df2.rename(columns = {'Name' : 'Local authority'}, inplace = True)
```

### 3 Prepare densities dataset

```python
df3.head()
```

```python
df3 = df3.drop(columns = ["Unnamed: 0"])
```

```python
df3 = df3.drop(columns = ["Lower Tier Local Authorities Code"])
```

### 4 Merge datasets

```python
df1.head(1)
```

```python
df1 = df1.drop(columns = ["Unnamed: 0"])
```

```python
# Merge df1 and df2
df1and2 = df1.merge(df2_wrangled, on = 'Local authority', indicator=True)
```

```python
df1and2.head(1)
```

```python
df1and2["_merge"].value_counts()
```

```python
df3.head(1)
```

```python
# Remove _merge flag column
df1and2 = df1and2.drop(columns = ["_merge"])
```

```python
# Merge df1and2 with df3
df_merged = df1and2.merge(df3, on = 'Local authority', indicator=True)
```

```python
df_merged.head(1)
```

```python
df_merged["_merge"].value_counts()
```

Merge successful!

```python
# Remove _merge flag column
df_merged = df_merged.drop(columns = ["_merge"])
```
